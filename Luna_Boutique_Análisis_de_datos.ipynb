```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
resumen = pd.DataFrame({
    "producto": ["Bolso", "Vestido", "Blusa", "Jeans", "Collar", "Aretes"],
    "ingresos": [240, 180, 150, 100, 75, 48],
    "ganancias": [140, 100, 90, 56, 50, 32]
})

resumen["margen_ganancia"] = (
    resumen["ganancias"] / resumen["ingresos"]
) * 100

resumen
```

```python
resumen

plt.figure(figsize=(10,5))

plt.bar(resumen["producto"], resumen["ingresos"])

plt.title("Ingresos por producto - Luna Boutique")
plt.xlabel("Productos")
plt.ylabel("Ingresos ($)")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(resumen["producto"], resumen["ganancias"])

plt.title("Ganancias por producto - Luna Boutique")
plt.xlabel("Productos")
plt.ylabel("Ganancias ($)")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(resumen["producto"], resumen["margen_ganancia"])

plt.title("Margen de ganancia por producto")
plt.xlabel("Productos")
plt.ylabel("Margen (%)")

plt.show()
```

```python
resumen.plot(
    x="producto",
    y=["ingresos", "ganancias"],
    kind="bar",
    figsize=(10, 6)
)

plt.title("Ingresos vs. ganancias - Luna Boutique")
plt.xlabel("Productos")
plt.ylabel("Cantidad ($)")
plt.xticks(rotation=45)
plt.legend(["Ingresos", "Ganancias"])
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

resumen.plot(
    x="producto", y="ingresos",
    kind="bar", ax=axes[0], legend=False
)
axes[0].set_title("Ingresos")

resumen.plot(
    x="producto", y="ganancias",
    kind="bar", ax=axes[1], legend=False
)
axes[1].set_title("Ganancias")

resumen.plot(
    x="producto", y="margen_ganancia",
    kind="bar", ax=axes[2], legend=False
)
axes[2].set_title("Margen de ganancia (%)")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Paleta de Luna Boutique
rosa = "#D98CA5"
malva = "#A98AAE"
lavanda = "#C9B5D5"
fondo = "#FFF6F8"
texto = "#594653"

# Preparar dashboard
fig = plt.figure(figsize=(17, 12), facecolor=fondo)

fig.suptitle(
    "LUNA BOUTIQUE",
    fontsize=27,
    fontweight="bold",
    color=texto,
    y=0.97
)

fig.text(
    0.5, 0.93,
    "SALES & PROFIT PERFORMANCE",
    ha="center",
    fontsize=12,
    color=malva
)

gs = fig.add_gridspec(2, 2, hspace=0.42, wspace=0.28)

ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[1, :])

def estilo(ax, titulo):
    ax.set_title(titulo, fontsize=15, color=texto,
                 fontweight="bold", pad=15)
    ax.set_facecolor("#FFFFFF")
    ax.grid(axis="x", alpha=0.15)
    ax.set_axisbelow(True)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)
    ax.tick_params(axis="both", colors=texto)

# Ingresos
barras = ax1.barh(
    resumen["producto"],
    resumen["ingresos"],
    color=rosa
)
estilo(ax1, "Ingresos por producto")
ax1.invert_yaxis()
ax1.bar_label(barras, padding=5, fmt="%.0f")
ax1.set_xlabel("Ingresos ($)", color=texto)

# Ganancias
barras = ax2.barh(
    resumen["producto"],
    resumen["ganancias"],
    color=malva
)
estilo(ax2, "Ganancias por producto")
ax2.invert_yaxis()
ax2.bar_label(barras, padding=5, fmt="%.0f")
ax2.set_xlabel("Ganancias ($)", color=texto)

# Margen de ganancia
barras = ax3.barh(
    resumen["producto"],
    resumen["margen_ganancia"],
    color=lavanda
)
estilo(ax3, "Margen de ganancia por producto")
ax3.invert_yaxis()
ax3.bar_label(
    barras,
    padding=5,
    fmt="%.2f%%"
)
ax3.set_xlabel("Margen (%)", color=texto)

plt.tight_layout(rect=[0.03, 0.03, 0.97, 0.90])
plt.show()
```

```python
import matplotlib.pyplot as plt

# Tipografías elegantes disponibles en Matplotlib
plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["DejaVu Serif"],
    "font.size": 11,
    "axes.titleweight": "normal",
    "axes.titlesize": 17,
    "axes.labelsize": 11,
    "figure.titlesize": 30,
    "text.color": "#594653",
    "axes.labelcolor": "#594653",
    "xtick.color": "#594653",
    "ytick.color": "#594653",
})

# Dashboard editorial
fig = plt.figure(figsize=(17, 12), facecolor="#FFF6F8")

fig.suptitle(
    "LUNA BOUTIQUE",
    fontsize=34,
    fontweight="normal",
    fontstyle="italic",
    color="#86566F",
    y=0.97
)

fig.text(
    0.5, 0.925,
    "THE ART OF FASHION  •  SALES REPORT",
    ha="center",
    fontsize=11,
    fontstyle="italic",
    color="#B58A9E"
)

gs = fig.add_gridspec(2, 2, hspace=0.42, wspace=0.28)

ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[1, :])

def boutique_style(ax, title):
    ax.set_title(title, pad=18, color="#86566F",
                 fontstyle="italic", fontsize=19)
    ax.set_facecolor("#FFFFFF")
    ax.grid(axis="x", alpha=0.12)
    ax.set_axisbelow(True)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)
    ax.spines["bottom"].set_color("#EAD8E1")

# Gráfica de ingresos
bars = ax1.barh(
    resumen["producto"], resumen["ingresos"],
    color="#D8A0B5", height=0.62
)
boutique_style(ax1, "Ingresos por producto")
ax1.invert_yaxis()
ax1.bar_label(bars, padding=6, fmt="%.0f")
ax1.set_xlabel("Ingresos ($)")

# Gráfica de ganancias
bars = ax2.barh(
    resumen["producto"], resumen["ganancias"],
    color="#B99ABF", height=0.62
)
boutique_style(ax2, "Ganancias por producto")
ax2.invert_yaxis()
ax2.bar_label(bars, padding=6, fmt="%.0f")
ax2.set_xlabel("Ganancias ($)")

# Gráfica de margen
bars = ax3.barh(
    resumen["producto"], resumen["margen_ganancia"],
    color="#C9A6B8", height=0.62
)
boutique_style(ax3, "Margen de ganancia")
ax3.invert_yaxis()
ax3.bar_label(bars, padding=6, fmt="%.2f%%")
ax3.set_xlabel("Margen (%)")

fig.text(
    0.5, 0.025,
    "LUNA BOUTIQUE  |  BUSINESS INTELLIGENCE",
    ha="center",
    fontsize=10,
    fontstyle="italic",
    color="#B58A9E"
)

plt.tight_layout(rect=[0.03, 0.06, 0.97, 0.89])
plt.show()
```
